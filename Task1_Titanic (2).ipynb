CHECK MISSING VALUE

```
df.isnull().sum()
```

HANDLE MISSING VALUES

```
df['Age'].fillna(df['Age'].mean(), inplace=True)
```

```
df['Embarked'].fillna(df['Embarked'].mode()[0], inplace=True)
```

```
df.drop('Cabin', axis=1, inplace=True)
```

```
df.isnull().sum()
```

```
df.duplicated().sum()
```

```
df.drop_duplicates(inplace=True)
```

```
df.duplicated().sum()
```

```
df.columns
```

```
df.dtypes
```

```
df['Age'] = df['Age'].astype(int)
```
